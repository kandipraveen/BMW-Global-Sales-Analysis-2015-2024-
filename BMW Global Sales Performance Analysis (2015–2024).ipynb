```python
# Calculate Q1, Q3, and IQR
Q1 = bmw['units_sold'].quantile(0.25)
Q3 = bmw['units_sold'].quantile(0.75)
IQR = Q3 - Q1

# Calculate outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Count outliers
outliers = bmw[(bmw['units_sold'] < lower_bound) | (bmw['units_sold'] > upper_bound)]
num_outliers = len(outliers)

print(f"Number of outliers in 'units_sold': {num_outliers}")
print(f"Lower bound for outliers: {lower_bound}")
print(f"Upper bound for outliers: {upper_bound}")
```

What is the distribution of sales (normal or skewed)?

```python
skewness = bmw['units_sold'].skew()
print(f"Skewness of 'units_sold': {skewness}")

if -0.5 <= skewness <= 0.5:
    print("The distribution of 'units_sold' is approximately symmetrical.")
elif skewness > 0.5:
    print("The distribution of 'units_sold' is positively (right) skewed.")
else:
    print("The distribution of 'units_sold' is negatively (left) skewed.")

print(f"Mean of 'units_sold': {bmw['units_sold'].mean()}")
print(f"Median of 'units_sold': {bmw['units_sold'].median()}")
```

##BMW Global Sales Performance Analysis (2015–2024)

## Project Overview

This project presents a comprehensive exploratory data analysis of BMW’s global sales performance from 2015 to 2024. The dataset consists of 1,000 structured records across 14 key business variables, including year, month, country, model, segment, engine type, price, marketing spend, dealership count, fuel price, GDP growth, interest rate, competition index, and units sold.

The objective of this analysis is to evaluate historical sales trends, identify top-performing markets and models, assess the influence of pricing and macroeconomic variables, and derive strategic business insights to support long-term growth.

The dataset was verified to be clean, with no missing values or duplicate entries, ensuring analytical reliability.

---

## Analytical Scope

The analysis covers the following key areas:

1. Sales trend evaluation and year-over-year growth analysis
2. Regional performance comparison
3. Model-wise and segment-wise sales assessment
4. Engine type growth evaluation
5. Competition intensity impact analysis
6. Revenue and pricing relationship assessment
7. Seasonal and quarterly performance trends
8. Statistical correlation analysis of key variables

---

## Key Findings

### 1. Sales Performance

Total units sold across the observed period amount to 633,192 units. The highest annual sales were recorded in 2015 (76,157 units), while the lowest occurred in 2016 (53,249 units). The period demonstrates notable volatility, including a decline of 30.08% in 2016 followed by a strong recovery of 30.99% in 2017. The year 2024 recorded a positive growth of 19.99%.

These findings indicate cyclical fluctuations but consistent recovery capability.

---

### 2. Regional Performance

China is the leading market in both sales volume and revenue generation:

* 77,488 units sold
* $6.07 billion in revenue
* 12.24% contribution to total global sales

In contrast, the United States recorded the lowest sales volume among major markets. Regional performance varies across years, indicating dynamic market conditions.

China remains BMW’s primary revenue driver.

---

### 3. Model and Segment Performance

The X3 model is the highest-selling and highest-revenue-generating model:

* 73,358 units sold
